# 6. Correlación de variables utilizando Python

Utiliza el conjunto de datos `load_breast_cancer` de Scikit-learn para identificar qué característica tiene la correlación lineal más fuerte (positiva o negativa) con la variable objetivo `target`.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer

# Cargar los datos
data = load_breast_cancer()

# Crear un DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)

# Agregar la variable objetivo (se invierte: 1 = maligno, 0 = benigno)
df["target"] = 1 - data.target

df.head()
```

**a) Calcula la correlación de Pearson entre cada característica y la variable `target`.**

```python
# corr() usa Pearson por defecto; nos quedamos con la columna de target
correlaciones = df.corr()["target"]
correlaciones
```

**b) Identifica la característica con mayor correlación en valor absoluto, excluyendo la correlación de `target` consigo misma.**

```python
# Quitamos target (su correlación consigo misma siempre es 1)
correlaciones = correlaciones.drop("target")

# abs() quita el signo para comparar la fuerza, sin importar si es positiva o negativa
correlaciones.abs().sort_values(ascending=False).head(5)
```

```python
mejor = correlaciones.abs().idxmax()
print("Característica con mayor correlación:", mejor)
print("Correlación:", round(correlaciones[mejor], 3))
```

**c) Indica si su correlación con `target` es positiva o negativa y explica brevemente qué significa.**

```python
if correlaciones[mejor] > 0:
    print("La correlación es positiva")
else:
    print("La correlación es negativa")
```

**Conclusión:**

La característica con la correlación lineal más fuerte con `target` es **`worst concave points`** (el valor más alto de puntos cóncavos del contorno de la célula), con una correlación de Pearson de **0.794**.

La correlación es **positiva**. Como en este ejercicio `target = 1 - data.target`, el valor 1 representa un tumor **maligno** y 0 uno **benigno**. Por lo tanto, mientras más puntos cóncavos tiene el contorno de la célula, más probable es que el tumor sea maligno. Un valor cercano a 0.8 indica una relación lineal fuerte, aunque la correlación por sí sola no demuestra causa.

Otras características muy cercanas son `worst perimeter` (0.783), `mean concave points` (0.777) y `worst radius` (0.776): las que describen el tamaño y la forma irregular de las células son las más relacionadas con el diagnóstico.
