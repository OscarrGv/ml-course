## Parte 1 — Lista de estudiantes

Idea clave: **filtrar con una máscara booleana** (`df[df["Gender"] == "M"]`) y luego aplicar `.mean()`, `.max()`, `.min()` a una columna.
Ojo: la estatura está en **cm**, así que 1.75 m = 175.

```python
df = pd.read_csv(DATA_DIR / "data_studentlist_en.csv")
df.head(5)
```

```python
hombres = df[df["Gender"] == "M"]
mujeres = df[df["Gender"] == "F"]

print("1) Altura promedio hombres:", hombres["Height"].mean())
print("2) Altura promedio mujeres:", mujeres["Height"].mean())
print("3) Peso promedio hombres:", hombres["Weight"].mean())
print("4) Peso promedio mujeres:", mujeres["Weight"].mean())
print("5) Mayor estatura hombres:", hombres["Height"].max())
print("6) Menor estatura mujeres:", mujeres["Height"].min())
```

Para combinar condiciones se usa `&` (y) o `|` (o), **cada condición entre paréntesis**. `isin()` es un atajo para varios `|` sobre la misma columna.

```python
print("7) Peso más ligero, hombres > 175 cm:", hombres[hombres["Height"] > 175]["Weight"].min())
print("8) Mayor peso, mujeres < 160 cm:", mujeres[mujeres["Height"] < 160]["Weight"].max())
print("9) Nota promedio sin ausencias:", df[df["Absence"] == "N"]["Grade"].mean())
print("10) Nota promedio con ausencias:", df[df["Absence"] == "Y"]["Grade"].mean())
print("11) Altura promedio grupo A o AB:", df[df["Bloodtype"].isin(["A", "AB"])]["Height"].mean())
print("12) Altura promedio hombres A o AB:", hombres[hombres["Bloodtype"].isin(["A", "AB"])]["Height"].mean())
print("13) Edad promedio, ausencias y nota >= 3:", df[(df["Absence"] == "Y") & (df["Grade"] >= 3)]["Age"].mean())
```
