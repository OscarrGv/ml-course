### 4. Responde las siguientes preguntas.

````python
df_cafeterias = pd.read_csv(DATA_DIR / "coffeeshop.csv", header="infer", na_values=[" "])
print(df_cafeterias.shape)
df_cafeterias.head(5)
````

**1) Genera una tabla de frecuencias para la variable 'yearOfStart' y visualízala por año.**
- **Ordena por año.**
- **Dibuja una gráfica de línea de 1997 a 2014. <= Pista: plt.xlim()**

````python
frecuencia_anual = df_cafeterias["Year of Start"].value_counts().sort_index()
frecuencia_anual
````

````python
frecuencia_anual.plot(kind="line")
plt.xlim(1997, 2014)
plt.title("Cafeterías abiertas por año")
plt.xlabel("Año de apertura")
plt.ylabel("Frecuencia")
plt.show()
````

**2) Divide los datos por el estado actual del negocio ('In' o 'Out'). Después, visualiza la tendencia anual de las frecuencias de la variable 'yearOfStart'.**
- **Ordena por año.**
- **Dibuja dos gráficas de líneas superpuestas de 1997 a 2014.**
- **Usa el objeto 'figure'.**

````python
abiertas = df_cafeterias[df_cafeterias["Current State"] == "In"]["Year of Start"].value_counts().sort_index()
cerradas = df_cafeterias[df_cafeterias["Current State"] == "Out"]["Year of Start"].value_counts().sort_index()

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)
ax.plot(abiertas, label="In (sigue abierta)")
ax.plot(cerradas, label="Out (cerrada)")
ax.set_xlim(1997, 2014)
ax.set_title("Cafeterías por año de apertura y estado actual")
ax.set_xlabel("Año de apertura")
ax.set_ylabel("Frecuencia")
ax.legend()
plt.show()
````

**3) A partir de los resultados de 1) y 2), ¿cómo describirías la tendencia en general?**

**Respuesta:** El número de cafeterías que abren cada año crece de forma constante desde 1997 y se acelera mucho a partir de 2008-2009: en 2014 abrieron más de 9,900. Las cafeterías antiguas se cerraron en su mayoría: en casi todos los años hasta 2009 hay más negocios cerrados ('Out') que abiertos ('In'). En cambio, desde 2010 la mayoría de las cafeterías nuevas siguen abiertas, y las cerradas bajan a partir de 2011. Esto tiene sentido porque los negocios más recientes han tenido menos tiempo para cerrar. En general, el mercado de las cafeterías está en plena expansión.
